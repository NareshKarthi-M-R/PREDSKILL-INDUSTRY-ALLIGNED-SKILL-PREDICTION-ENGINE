```python
import pandas as pd
import pickle

# Load dataset
df = pd.read_csv("C:/Users/Lekas/Downloads/PredSkill2/data/job_posting.csv")

df = df.dropna()

# Normalize skills
df['skills'] = df['skills'].str.lower()
df['skills'] = df['skills'].apply(lambda x: [s.strip() for s in x.split(",")])

# -----------------------------
# CLEAN SKILLS BASED ON ROLE
# -----------------------------
def clean_skills_by_role(role, skills):
    role = role.lower()

    if "data" in role or "analyst" in role:
        allowed = [
            "python","sql","pandas","numpy","matplotlib",
            "data analysis","statistics","excel","power bi"
        ]

    elif "ai" in role or "machine learning" in role:
        allowed = [
            "python","machine learning","deep learning",
            "tensorflow","pytorch","nlp"
        ]

    elif "frontend" in role or "backend" in role or "web" in role:
        allowed = [
            "html","css","javascript","react","node.js","mongodb"
        ]

    elif "cloud" in role or "devops" in role:
        allowed = [
            "aws","docker","kubernetes","ci/cd"
        ]

    else:
        return skills

    return [s for s in skills if s in allowed]

# -----------------------------
# BUILD ROLE SKILLS
# -----------------------------
role_skills = {}

for _, row in df.iterrows():
    role = row['job_title']
    skills = row['skills']

    cleaned = clean_skills_by_role(role, skills)

    if role not in role_skills:
        role_skills[role] = set()

    role_skills[role].update(cleaned)

# Remove empty roles
role_skills = {k: v for k, v in role_skills.items() if len(v) > 0}

# Limit size
for role in role_skills:
    role_skills[role] = set(list(role_skills[role])[:15])

# Save
pickle.dump(role_skills, open("role_skills.pkl","wb"))
```
